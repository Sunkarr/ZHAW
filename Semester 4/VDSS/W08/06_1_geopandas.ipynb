## File formats for geo data
Some common file formats for geo data:
* Shapefiles: .shp, .shx, .dbf (folders with collections of such files are often compressed into a .zip)
* GPS Exchange Format: .gpx
* Keyhole Markup Language (KML): .kmz
* MapInfo TAB (proprietary): .tab, .dat, .ind

Coordinates are always given in the order longitude, latitude.

## Managing Coordinate Systems and Projections
https://kodu.ut.ee/~kmoch/geopython2021/L2/crs-projections.html

A **geographic coordinate system** (GCS) uses a ellipsoidal surface to define locations on the Earth. There are three parts to a geographic coordinate system:
* A datum - an ellipsoidal (spheroid) model of the Earth to use. Common datums include 
    * WGS84 (used in GPS)
    * NAD83 (North American Datum)
    * OSGB36 (for Great Britain)
    * ETRS89 (the European Datum)
* A prime meridian
* Angular unit of measure (degrees or grads)

Both latitude and longitude are typically represented in
* in Degrees, Minutes, Seconds (DMS), for example, 58° 23′ 12′ ′N, 26° 43′ 21′ ′E
* or Decimal Degrees (DD) used by computers and stored as float data type, for example, 58.38667 and 26.7225

**Projected coordinate systems** (PCS) define a flat 2D Cartesian surface. Unlike a geographic coordinate system, a projected coordinate system has constant lengths, angles, and areas across the two dimensions. A projected coordinate system is always based on a geographic coordinate system that references a specific datum. Projected Coordinate Systems consist of:
* Geographic Coordinate System
* Projection Method
* Projection Parameters (standard points and lines, Latitude of Origin, Longitude of Origin, False Easting, False Northing etc)
* Linear units (meters, kilometers, miles etc)

![What your favorite map projection says about you](https://imgs.xkcd.com/comics/map_projections.png)

````python
europe = gpd.read_file(os.path.join(data_folder, 'Europe_borders', 'Europe_borders.shp'))
````

The projection loaded with the data:

````python
europe.crs
````

To convert, i.e. reproject, those geometries into geopandas provides the `.to_crs()`-function. Both ways are equivalent:
* my_geoseries.set_crs("EPSG:4326")
* my_geoseries.set_crs(epsg=4326)

For example, convert to the Lambert Azimuthal Equal Area projection (EPSG: 3035), which is the recommended projection for Europe: 

````python
europe_proj = europe.copy()

europe_proj = europe_proj.to_crs(epsg=3035) # europe_proj = europe_proj.to_crs(crs='EPSG:3035')
````

````python
europe_proj.crs
````

````python
fig, axs = plt.subplots(1, 2, figsize=(12,6))

# Plot the WGS84
europe.plot(ax=axs[0], facecolor='gray')

# Add title
axs[0].set_title("WGS84 CRS", fontsize=16)
axs[0].set_ylabel('Latitude (degree)', fontsize=14)
axs[0].set_xlabel('Longitude (degree)', fontsize=14)

# Plot the one with ETRS-LAEA projection
europe_proj.plot(ax=axs[1], facecolor='tab:blue')

# Add title
axs[1].set_title("ETRS Lambert Azimuthal Equal Area projection", fontsize=16)
axs[1].set_ylabel('Northing (metre)', fontsize=14)
axs[1].set_xlabel('Easting (metre)', fontsize=14)

# Remove empty white space around the plot
plt.tight_layout()
plt.show()
````

## Touching and Intersecting
The difference is:
* If objects intersect, the boundary and interior of an object needs to intersect in any way with those of the other.

* If an object touches the other one, it is only necessary to have (at least) a single point of their boundaries in common but their interiors shoud NOT intersect.

````python
gdf_proj[gdf_proj['name']=='TD building'].iloc[0]
````

````python
gdf_proj.iloc[6]
````

````python
gdf_proj['geometry'].intersects(gdf_proj.iloc[6]['geometry'])
````

````python
gdf_proj['geometry'].touches(gdf_proj.iloc[6]['geometry'])
````
