```python
f_x = lambda x: x**3 + 4*x + 1
s = []
for i in range(11):
    s.append(f_x(i) % 11)
    print(f"f({i}) mod 11 = {s[-1]}")
```

```python
quad_res = lambda s, p: s**((p-1)/2) % p
p = 11
q_res = []
for i in s:
    q_res.append(quad_res(i, p))
    print(f"Quadratic residue of {i} mod {p} is {q_res[-1]}")
```
