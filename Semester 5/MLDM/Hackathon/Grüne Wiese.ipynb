```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import mplfinance as mpf
```

```python
crypto = pd.read_csv('/Users/jonas/Documents/GitHub/ZHAW/Semester 5/MLDM/Hackathon/data/crypto_time_series.csv')

# separate after "coin"
df = crypto[crypto['coin'] == 'BTC'].copy()
df['date'] = pd.to_datetime(df['date'])
```

```python
# Calculate the daily average price
btc = df.copy()
btc['avg_price'] = (df['open'] + df['high'] + df['low'] + df['close']) / 4
btc.drop('low', axis=1, inplace=True)
btc.drop('open', axis=1, inplace=True)
btc.drop('high', axis=1, inplace=True)
btc.drop('close', axis=1, inplace=True)
btc.drop('market_cap', axis=1, inplace=True)
btc.drop('coin', axis=1, inplace=True)
btc
```

```python
# calculate the x daily returns
x_days = 30
btc[f'return_{x_days}d'] = btc['avg_price'].pct_change(x_days) * 100
btc.dropna(inplace=True)
btc
```
