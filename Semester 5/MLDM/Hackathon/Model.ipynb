```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/Users/jonas/Documents/GitHub/ZHAW/Semester 5/MLDM/Hackathon/data/crypto_time_series.csv')

# Parse date and sort
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date').reset_index(drop=True)

# Optional: drop rows with missing key prices (e.g. NaN open)
df = df.dropna()

df
```

```python
df["label"] = 0  # 0 = none
df.loc[df["is_top"] == 1, "label"] = 1   # 1 = local top
df.loc[df["is_bottom"] == 1, "label"] = 2  # 2 = local bottom

# Edge case: both 1 (flat area), set to none or drop
conflict_idx = (df["is_top"] == 1) & (df["is_bottom"] == 1)
df.loc[conflict_idx, "label"] = 0
```
