# Load Dataset

You can load the dataset directly with pandas.

```python
import pandas as pd

df = pd.read_csv('crypto_time_series.csv')
```

## Content
Contains historical data for 4 different crypto currencies.

## Currencies
- BTC: Bitcoin
- BITS: Global X Blockchain & Bitcoin Strategy ETF
- ETH: Ethereum
- LTC: Litecoin

## Columns
- Date - the day of recorded values
- Open - the opening price (in USD)
- High - the highest price (in USD)
- Low - the lowest price (in USD)
- Close - the closing price (in USD)
- Volume - total exchanged volume (in USD)
- Market Cap - the total market capitalization for the coin (in USD)
- coin - the name of the coin
