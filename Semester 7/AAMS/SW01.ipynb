```python
from matplotlib import pyplot as plt

plt.style.use('seaborn')

def x_t (x, r):
    return r * x * (1 - x)

for r in [2.5, 3.2, 3.5, 3.8]:
    x = 0.1
    x_values = []
    for i in range(100):
        x = x_t(x, r)
        x_values.append(x)
    plt.plot(x_values, label=f'r={r}')
    plt.title(f'r={r}')
```
