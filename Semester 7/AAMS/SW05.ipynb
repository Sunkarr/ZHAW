```python
import numpy as np
import math as m
```

```python
x = np.array([[-1, 2], [2, -2]])
y = np.array([[0.5, -1.5], [1, -1]])
```

```python
# calc eigenvalues and eigenvectors
eigenvalues_x, eigenvectors_x = np.linalg.eig(x)
eigenvalues_y, eigenvectors_y = np.linalg.eig(y)

print("Eigenvalues of x:", eigenvalues_x)
print("Eigenvectors of x:\n", eigenvectors_x)
print("Eigenvalues of y:", eigenvalues_y)
print("Eigenvectors of y:\n", eigenvectors_y)
```
